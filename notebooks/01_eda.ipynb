```python

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
df[df['Class']==0]['Amount'].plot(kind='hist', bins=50, ax=ax[0], title='Non-fraud Amount')
df[df['Class']==1]['Amount'].plot(kind='hist', bins=50, ax=ax[1], title='Fraud Amount')
ax[0].set_yscale('log')
ax[1].set_yscale('log')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
df[df['Class']==0]['Time'].plot(kind='hist', bins=50, alpha=0.7, label='Non-fraud')
df[df['Class']==1]['Time'].plot(kind='hist', bins=50, alpha=1, label='Fraud')
plt.legend()
plt.show()
```

```python
cols = ['Time', 'Amount']
tables = {}
for col in cols:
    table = df.groupby(target).agg({col : ['mean', 'median', 'min', 'max']})
    tables[col] = table
    print(f"Distribution for {col}: ")
    print(tables[col])
```

- Amount: fraud (class 1) has a much higher mean (123.87) but lower median (9.82) than non-fraud (mean 88.41, median 22.00) — fraud amounts are more spread out/skewed, with a few large fraud amounts pulling the mean up while most fraud transactions are actually small.

- Time: fraud skews slightly earlier (mean 80450 vs 94835), but the ranges overlap a lot — probably not a strong standalone signal.

V1–V28 individually: they're anonymized PCA outputs — there's no real-world meaning to interpret ("V14 is high" tells you nothing about actual behavior). Digging into them one by one wastes time for zero insight.
